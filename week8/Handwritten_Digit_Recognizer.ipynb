# Handwritten Digit Recognizer (MNIST)

## EDP Internship — Machine Learning Project

This project focuses on recognizing handwritten digits from 0 to 9 using the MNIST dataset.

The project covers image preprocessing, normalization, KNN classification, neural networks, model evaluation, accuracy improvement, and visualization.

## Objectives

- Understand image data preprocessing
- Normalize MNIST image data
- Train a KNN baseline model
- Train a Neural Network model
- Improve model accuracy
- Analyze misclassified digits
- Visualize model performance
- Document the complete Machine Learning workflow

## Dataset

The MNIST dataset contains handwritten digit images representing digits from 0 to 9.

Each image has:

- 28 × 28 pixels
- Grayscale format
- One digit label from 0 to 9

The dataset is commonly used for handwritten digit classification.

## KNN Model

K-Nearest Neighbors (KNN) was used as the baseline model for handwritten digit classification.

The model was trained on the MNIST dataset and evaluated on the test data.

### KNN Accuracy

**94.63%**

## Neural Network Model

A Neural Network model was trained to improve the handwritten digit classification performance.

The model was trained using the MNIST dataset and evaluated on the test data.

### Improved Model Accuracy

**98.02%**

## Model Performance

| Model | Accuracy |
|---|---:|
| KNN | 94.63% |
| Improved Neural Network | 98.02% |

The Neural Network achieved higher accuracy than the KNN baseline model.

## Model Evaluation

The models were evaluated using:

- Accuracy
- Confusion Matrix
- Actual vs Predicted values
- Misclassified images
- Data visualization

These evaluation methods helped analyze the model's performance and identify incorrectly classified digits.

## Confusion Matrix

A confusion matrix was used to evaluate the classification performance of the model.

It shows the number of correctly and incorrectly predicted digits for each class from 0 to 9.

The confusion matrix helps identify which digits are more frequently confused with one another.

## Misclassified Images

Misclassified images were analyzed to understand where the model made incorrect predictions.

The analysis compares:

- Actual digit
- Predicted digit
- Image of the handwritten digit

This helps identify difficult handwritten patterns and understand the limitations of the model.
